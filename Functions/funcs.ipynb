```python
def addition(a,b):
    c = a + b
    return c

print(addition(2,3))
```

```python
def evenOdd(n):
    if n % 2 == 0: # If True This runs
        return "Even"
    else: # If False, this will run
        return "Odd"

nums = [2,67,32,577,13,4,4,623,45,57,78,23,45,687,23,46,57,68,4,57,68,797,934,6,79,2,46,7,8,86]
for i in nums:
    print(evenOdd(i))
```

### Declaring an Empty Dict

#### Syntax of the Dictionary 
#### dict_name = {
    "key1" : "value1",
    "key2" : "value2"
}

```python
vaishnavi_dict = {
    "name" : "Vaishnavi",
    "age" : 20,
    "gender" : "Female",
    "marks" : 90.5
}
print(vaishnavi_dict)
```

```python
sirisha_dict = {
    "name" : "Sirisha",
    "age" : 21,
    "gender" : "Female",
    "marks" : 95.0,
    "hobbies" : ["Reading", "Cooking", "Traveling"]
}
print(sirisha_dict)
```

#### Using Dicts to Record data VAlues in Categories

```python
temps = {
    "Monday" : 45.7,
    "Tuesday" : 23.9,
    "Wednesday" : 35.1,
    "Thursday" : 38.2,
    "Friday" : 33.9,
    "Saturday" : 36.1
}
for day,temp in enumerate(temps):
    print(f"The Temperature on {temp} is {temps[temp]}")
```

#### Accessing Elements In a Dictionary
#### dict_name["key"] ==> It gives the Value Assigned to That Key

```python
print(temps["Monday"])
print(sirisha_dict["hobbies"])
print(vaishnavi_dict["name"])
```

#### Chanhing Elements in Dict

```python
temps["Thursday"] = 28.2
print(temps["Thursday"])
```

#### Removing Items in a Dictionary

```python
thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
print(thisdict)
thisdict.pop("model")
print(thisdict)
```
